```python
import os
import pandas as pd
import random
```

```python
# 1. Merge JSONL files and shuffle data
def merge_and_shuffle(folder_path, output_file):
    data = []
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".jsonl"):
            file_path = os.path.join(folder_path, file_name)
            with open(file_path, 'r', encoding='utf-8') as file:
                for line in file:
                    data.append(line.strip())
    random.shuffle(data)
    with open(output_file, 'w', encoding='utf-8') as file:
        file.write("\n".join(data))
    print(f"Merged and shuffled data saved to {output_file}")
```

```python
# 2. Update specific text in the merged JSONL file
def update_text(file_path, old_text, new_text):
    with open(file_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    updated_lines = [line.replace(old_text, new_text) for line in lines]
    with open(file_path, 'w', encoding='utf-8') as file:
        file.writelines(updated_lines)
    print(f"Updated text in {file_path}")
```

```python
# 3. Remove duplicate rows from the merged JSONL file
def remove_duplicates(file_path):
    df = pd.read_json(file_path, lines=True)
    print("Duplicate rows before removal:")
    print(df[df.duplicated()])
    df.drop_duplicates(inplace=True)
    print("Duplicate rows after removal:")
    print(df[df.duplicated()])
    df.to_json(file_path, orient='records', lines=True, force_ascii=False)
    print(f"Duplicates removed and data saved to {file_path}")
```

```python

# 4. Split the merged JSONL file into smaller files
def split_file(input_file, output_folder, num_splits=5):
    with open(input_file, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    num_lines = len(lines)
    lines_per_file = num_lines // num_splits
    remainder = num_lines % num_splits
    os.makedirs(output_folder, exist_ok=True)
    start = 0
    for i in range(num_splits):
        end = start + lines_per_file + (1 if i < remainder else 0)
        output_file = os.path.join(output_folder, f"split_{i + 1}.jsonl")
        with open(output_file, 'w', encoding='utf-8') as out_file:
            out_file.writelines(lines[start:end])
        print(f"File {output_file} created with {end - start} lines.")
        start = end
```

```python
# Main execution
if __name__ == "__main__":
    # Paths
    folder_path = "c:\\Users\\84793\\Desktop\\Code\\ML\\FinalProject\\PreData"
    output_file = "c:\\Users\\84793\\Desktop\\Code\\ML\\FinalProject\\merged_data.jsonl"
    split_output_folder = "c:\\Users\\84793\\Desktop\\Code\\ML\\FinalProject\\SplitData"

    # Step 1: Merge and shuffle data
    merge_and_shuffle(folder_path, output_file)

    # Step 2: Update specific text
    update_text(
        output_file,
        "Bạn là chuyên gia dinh dưỡng AI",
        "Bạn là chuyên gia dinh dưỡng AI. Hãy đưa ra DUY NHẤT MỘT thực đơn phù hợp."
    )

    # Step 3: Remove duplicate rows
    remove_duplicates(output_file)

    # Step 4: Split the file into smaller files
    split_file(output_file, split_output_folder)
```
